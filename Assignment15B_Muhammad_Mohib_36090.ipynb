```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("student_tea_snacks.csv")

# Q1
tea_counts = df["tea_type"].value_counts()
print(tea_counts)
print("Most preferred tea type:", tea_counts.idxmax())

# Q2
print(df.groupby("gender")["cups_per_day"].mean().round(2))

# Q3
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="snack_choice", order=df["snack_choice"].value_counts().index)
plt.title("Snack Choices")
plt.xlabel("Snack")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()

# Q4
print(pd.crosstab(df["age_group"], df["tea_type"]))
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="age_group", hue="tea_type", order=sorted(df["age_group"].unique()))
plt.title("Tea Preference by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()

# Q5
print("Students who don't drink tea:", (df["prefers_tea"] == "No").sum())
```
